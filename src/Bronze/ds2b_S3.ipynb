```python
from common_utils.logging import get_logger
logger=get_logger("ds2b_S3")
```

```python

import pyspark.sql.functions as F

logger.info("raw path defined ")

rawpath="/Volumes/retail_data_platform/bronze/raw_data/S3_bucket/2026-10-28/"

logger.info("reading data from %s",rawpath)

df= spark.read.format("parquet")\
    .option("inferSchema", "true")\
    .load(rawpath)

logger.info("auditing coloumns")

df=df.withColumn("last_update_ts",F.current_timestamp())\
 .withColumn("File_path",F.col("_metadata.file_path"))

logger.info("sample data %s",df.show(5))

logger.info("target path defined")

target_path="retail_data_platform.bronze.S3_sales"

logger.info(" writing data to %s",target_path)

df.write.format("delta")\
    .mode("overwrite")\
    .saveAsTable(target_path)


logger.info("data written to %s",target_path)
```
